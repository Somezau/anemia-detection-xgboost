```python
import pandas as pd

data = [
    # rasio, n_estimators, accuracy, precision, recall, f1
    ['60:40', 50, 0.906854, 0.888446, 0.899194, 0.893788],
    ['60:40', 70, 0.913884, 0.890196, 0.915323, 0.902584],
    ['60:40', 100, 0.938489, 0.917647, 0.943548, 0.930417],

    ['70:30', 50, 0.934426, 0.895000, 0.962366, 0.927461],
    ['70:30', 70, 0.932084, 0.879227, 0.978495, 0.926209],
    ['70:30', 100, 0.941452, 0.888889, 0.989247, 0.936387],

    ['80:20', 50, 0.908772, 0.882812, 0.911290, 0.896825],
    ['80:20', 70, 0.933333, 0.900763, 0.951613, 0.925490],
    ['80:20', 100, 0.933333, 0.894737, 0.959677, 0.926070],
]

df = pd.DataFrame(data, columns=[
    'Rasio', 'n_estimators', 'Accuracy', 'Precision', 'Recall', 'F1-Score'
])

df
```

```python
import matplotlib.pyplot as plt

for rasio in df['Rasio'].unique():
    subset = df[df['Rasio'] == rasio]
    plt.plot(subset['n_estimators'], subset['Accuracy'], marker='o', label=rasio)

plt.xlabel('n_estimators')
plt.ylabel('Accuracy')
plt.title('Perbandingan Accuracy pada Berbagai Rasio Data')
plt.legend()
plt.show()
```

```python
for rasio in df['Rasio'].unique():
    subset = df[df['Rasio'] == rasio]
    plt.plot(subset['n_estimators'], subset['F1-Score'], marker='o', label=rasio)

plt.xlabel('n_estimators')
plt.ylabel('F1-Score')
plt.title('Perbandingan F1-Score pada Berbagai Rasio Data')
plt.legend()
plt.show()
```

```python
for rasio in df['Rasio'].unique():
    subset = df[df['Rasio'] == rasio]
    plt.plot(subset['n_estimators'], subset['Precision'], marker='o', label=rasio)

plt.xlabel('n_estimators')
plt.ylabel('Precision')
plt.title('Perbandingan Precision pada Berbagai Rasio Data')
plt.legend()
plt.show()
```

```python
for rasio in df['Rasio'].unique():
    subset = df[df['Rasio'] == rasio]
    plt.plot(subset['n_estimators'], subset['Recall'], marker='o', label=rasio)

plt.xlabel('n_estimators')
plt.ylabel('Recall')
plt.title('Perbandingan Recall pada Berbagai Rasio Data')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

for rasio in df['Rasio'].unique():
    subset = df[df['Rasio'] == rasio]

    plt.figure(figsize=(6,4))

    for metric in metrics:
        plt.plot(subset['n_estimators'], subset[metric], marker='o', label=metric)

    plt.title(f'Perbandingan Metrik Evaluasi (Rasio {rasio})')
    plt.xlabel('n_estimators')
    plt.ylabel('Nilai')
    plt.legend()
    plt.show()
```
